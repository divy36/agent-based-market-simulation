# Market simulation walkthrough
Run the package; inspect the saved tables and figures.

```python
from market_sim.agents import ZICAgent
from market_sim import DoubleAuctionMarket, SimulationPipeline
buyers = [ZICAgent(i, True, 150, 0, 200) for i in range(4)]
sellers = [ZICAgent(i + 4, False, 50, 0, 200) for i in range(4)]
market = DoubleAuctionMarket(buyers, sellers, random_seed=42)
market.run(100)
SimulationPipeline(market).summary_stats()
```
